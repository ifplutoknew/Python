```python

import matplotlib.pyplot as plt

# Strain rate (mm/min)
strain_rate = [2, 20, 200, 2000]

# Toe length (mm)
curly_hair = [3.2554, 2.2818, 1.5864, 1.5400]
straight_hair = [1.5400, 1.5864, 2.7455, 1.6791]

# Create the graph
plt.figure(figsize=(9, 6))

# Plot curly hair
plt.plot(
    strain_rate,
    curly_hair,
    marker='o',
    linestyle='-',
    label='Curly hair'
)

# Plot straight hair
plt.plot(
    strain_rate,
    straight_hair,
    marker='s',
    linestyle='-',
    label='Straight hair'
)

# Add titles and labels
plt.title('Effect of Strain Rate on Toe Length')
plt.xlabel('Strain Rate (mm/min)')
plt.ylabel('Toe Length (mm)')

# Use a logarithmic scale for the x-axis
plt.xscale('log')

# Add grid and legend
plt.grid(True, which='both', linestyle='--', alpha=0.6)
plt.legend()

# Display the graph
plt.tight_layout()
plt.show()
```
